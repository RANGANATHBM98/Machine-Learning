```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)
x=np.linspace(0,10,20).reshape(-1,1)
y=2*x.squeeze()+3+np.random.normal(0,2,20)
print(x)
print(y)
print(x.shape)
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
y_pred=model.predict(x)
plt.figure(figsize=(8,6))
plt.scatter(x,y,color='red',label='training data')
plt.plot(x,y_pred,color='blue',linewidth=3,label='linear Regression')
plt.title("simple linear regression plotting")
```

```python
poly=PolynomialFeatures(degree=5)
x_poly=poly.fit_transform(x)
```

```python
print("Original x shape:",x.shape)
print("polynomial x shape:",x_poly.shape)
```

```python
print("\n first 5 row of original x shape:")
print(x[:5].flatten())
print("polynomial first 5 rows")
print(np.round(x_poly[:5],4))
```

```python
degrees = [2,5,10,15]
x_test = np.linspace(0,10,300).reshape(-1,1)
fig,axes = plt.subplots(nrows=2,ncols=2,figsize=(14,14))
axes = axes.ravel()
for i, degree in enumerate(degrees):
    poly = PolynomialFeatures(degree=degree)
    x_poly = poly.fit_transform(x)
    model = LinearRegression()
    model.fit(x_poly,y)
    y_pred = model.predict(poly.transform(x_test))
    axes[i].scatter(x,y,color="red",label="Training_data")
    axes[i].plot(x_test,y_pred,label=f"Degree {degree}")
    axes[i].set_title(f"Polynomial Regression Degree {degree}")
    axes[i].legend()
plt.show()
```
